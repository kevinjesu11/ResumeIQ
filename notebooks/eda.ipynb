# 📊 ResumeIQ — Exploratory Data Analysis


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re

df = pd.read_csv('../data/resumes.csv')
print('Shape:', df.shape)
df.head(2)
```

```python
df['text_length'] = df['resume_text'].apply(lambda x: len(str(x).split()))
print('Average word count:', df['text_length'].mean().round(0))
df['text_length'].hist(bins=30, color='#C8541A', edgecolor='white')
plt.title('Resume Word Count Distribution')
plt.xlabel('Word Count')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
roles = {
    'Java Developer': ['java','spring','hibernate','j2ee'],
    'Business Analyst': ['business analyst','requirements','stakeholder','brd'],
    'Project Manager': ['project manager','pmp','agile','scrum'],
    'DevOps': ['docker','kubernetes','jenkins','terraform']
}

for role, kws in roles.items():
    count = df['resume_text'].apply(lambda x: any(k in str(x).lower() for k in kws)).sum()
    print(f'{role}: {count} resumes ({count/len(df)*100:.1f}%)')
```

```python
print('Shortlisted value counts:')
print(df['shortlisted'].value_counts())
df['shortlisted'].value_counts().plot(kind='bar', color=['#EF4444','#10B981'], edgecolor='white')
plt.title('Shortlisted vs Not Shortlisted')
plt.xticks([0,1], ['Not Shortlisted','Shortlisted'], rotation=0)
plt.tight_layout()
plt.show()
```
